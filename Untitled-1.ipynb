```python
import os
import numpy as np
import pandas as pd

# Verify working directory
print("Current Working Directory:", os.getcwd())
```

```python
raw_path = "../data/raw/diabetic_data.csv" if os.path.exists("../data/raw/diabetic_data.csv") else "data/raw/diabetic_data.csv"

df = pd.read_csv(raw_path)
print(f"Initial raw shape: {df.shape}")
df.head()
```

```python
# Convert '?' into standard NaN values
df.replace("?", np.nan, inplace=True)

# Inspect columns with missing values
missing_counts = df.isna().sum()
missing_summary = missing_counts[missing_counts > 0].sort_values(ascending=False)
print("Columns with missing values:")
print(missing_summary)
```

```python
initial_count = len(df)
df.drop_duplicates(inplace=True)
print(f"Removed {initial_count - len(df)} identical duplicate rows. Current rows: {len(df)}")
```

```python
# Deduplicate on patient_nbr to prevent data leakage across repeat hospitalizations
if "patient_nbr" in df.columns and "encounter_id" in df.columns:
    df.sort_values(by="encounter_id", inplace=True)
    before_dedup = len(df)
    df.drop_duplicates(subset=["patient_nbr"], keep="first", inplace=True)
    print(f"Removed {before_dedup - len(df)} repeat visits. Unique patients remaining: {len(df)}")
else:
    print("Columns patient_nbr or encounter_id not found.")
```

```python
# IDs: 11 (Expired), 13 (Hospice/home), 14 (Hospice/facility), 19/20/21 (Expired)
terminal_ids = [11, 13, 14, 19, 20, 21]

if "discharge_disposition_id" in df.columns:
    before_filter = len(df)
    df = df[~df["discharge_disposition_id"].isin(terminal_ids)]
    print(f"Removed {before_filter - len(df)} deceased/hospice records. Remaining rows: {len(df)}")
else:
    print("discharge_disposition_id column not found.")
```

```python
# Drop non-predictive IDs and columns with excessive missingness (>40%)
cols_to_drop = ["encounter_id", "patient_nbr", "weight", "payer_code", "medical_specialty"]

existing_drop_cols = [col for col in cols_to_drop if col in df.columns]
df.drop(columns=existing_drop_cols, inplace=True)
print(f"Dropped columns: {existing_drop_cols}")
print(f"Remaining columns count: {df.shape[1]}")
```

```python
output_path = "../data/cleaned_step1.csv" if os.path.exists("../data") else "data/cleaned_step1.csv"

os.makedirs(os.path.dirname(output_path), exist_ok=True)
df.to_csv(output_path, index=False)

print(f"Cleaned dataset successfully saved to: {output_path}")
print(f"Final cleaned shape: {df.shape}")
df.head()
```
